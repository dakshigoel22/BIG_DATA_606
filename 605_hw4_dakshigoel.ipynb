#1. What is Neo4j?

**○ Explain what kind of datastore it is, and when it is most useful.**

Neo4j is a native property graph database which means nodes represent entities, relations represents connections and properties represent key-value data stored.

It is most useful when the data is highly connected and the relations matter.

Common use cases include :
1. Recommendation systems
2. Fraud detection
3. Social Networks


**What does Neo4j do natively and why does the idea of “native” matter?**

Native here means that Neo4j is built from the ground up for graphs and not layered on top of another model.

Neo4j natively supports:
1. Graph storage
2. Graph traversal
3. Graph query language
4. Index-free adjacency

Native matters because the relations are physically stored and directly linked . It also makes traversals fast and predictable. Query complexity depends on graph shape, not dataset size

#Graph Design

#Representation of the data as a property graph model
The deepest reason to use a property graph for this domain is that the interesting queries are multi-hop traversals.  To represent the business model (i.e. Skillify) as a Neo4j property graph:

1.  All the entities will become the nodes of the graph. The user node becomes (:USER) node.  The columns become properties stored directly on each node, so a User node might look like
```
{user_id: "u1", name: "Alice", role: "student"}
```

2. Foreign keys become Relationships (Edges). All the foreign key in the relational schema becomes an actual traversable edge. For ex, a choreographer_id in WorkshopEvent or user making a booking

```
(:User)-[:MADE]->(:Booking)
```

3. Juntion tables becomes the labelled edges with property : The M:M relationship of a user to a workshop , would require a junction table. But in Neo4j ,that table disappears entirely and becomes a single relationship:

```
(:User)-[:REGISTERED_FOR {registered_at: "...", status: "confirmed"}]->(:WorkshopEvent)

```







■ What properties might it make sense to have on edges (relationships between entities) ?

We put a property on an edge when it describes the circumstance of that specific connection not when it describes either of the nodes themselves.

From the Skillify business models , strong candidates are:

[:REGISTERED_FOR] should carry registered_at and status - because the same user can cancel and re register,creating two different edges with different timestamps and statuses.

[:HOSTED_AT] should carry date and start_time - the same studio hosts many different events on different days. The date is a property of this particular pairing of event and studios, not of the studio itself.

[:PAID_BY] should carry paid_at and possibly payment_method - the moment of payment is about this specific settlement of this specific booking.

[:GAVE] (User gives a Review) could carry reviewed_at



■ What should nodes look like?

Nodes in Neo4j are Json objects attached to a label
- (:User) node holds everything intrinsic to that person: {user_id, name, role}
- (:WorkshopEvent) node holds {event_id, title, price, level, dance_style}
- (:Booking) node is an interesting case as relationally booking of often a junction table. In graph table, I will keep it as a real node as it has its own identity (booking_id) and connects to both payment and workshopEvents

````python
from neo4j import GraphDatabase

uri = "bolt://localhost:7687"
username = "neo4j"
password = "neo4_pass"

driver = GraphDatabase.driver(uri, auth=(username, password))
driver.verify_connectivity()
````

````python
def run_query(tx, query):
        result = tx.run(query)
        return result.data()


with driver.session() as session:
  query= "SHOW DATABASES yield name;"
  result = session.execute_read(run_query, query)
  print(result)
````

````python
with driver.session() as session:
  session.run("CREATE (:User {user_id: 'u1', name: 'Alice Chen', role: 'student'})")
  session.run("CREATE (:User {user_id: 'u2', name: 'Marcus webb', role: 'student'})")
  session.run("CREATE (:User {user_id: 'u3', name: 'Bob', role: 'student'})")
  session.run("CREATE (:ChoreographerProfile {choreographer_id: 'cp1', name: 'Sofia Reyes', rating: 4.8})")
  session.run("CREATE (:ChoreographerProfile {choreographer_id: 'cp2', name: 'James Rees', rating: 3.2})")
  session.run("CREATE (:Studio {studio_id: 's1', name: 'Rhythm Space', capacity: 30})")
  session.run("CREATE (:WorkshopEvent {event_id: 'e1', title: 'Afrobeats Fundamentals', price: 25.0})")
  session.run("CREATE (:Booking {booking_id: 'b1', status: 'confirmed'})")
  session.run("CREATE (:Booking {booking_id: 'b2', status: 'initated'})")
  session.run("CREATE (:Payment {payment_id: 'p1', amount: 25.0, payment_status: 'completed'})")
  session.run("CREATE (:Review {review_id: 'r1', rating: 5, comment: 'Amazing class!'})")
  print("Nodes created.")
````

````python
with driver.session() as session:
    session.run("""
        MATCH (u:User {user_id: 'u1'}), (b:Booking {booking_id: 'b1'})
        CREATE (u)-[:MADE {booked_at: '2024-11-01'}]->(b)
    """)
    session.run("""
        MATCH (b:Booking {booking_id: 'b1'}), (e:WorkshopEvent {event_id: 'e1'})
        CREATE (b)-[:FOR_EVENT]->(e)
    """)
    session.run("""
        MATCH (b:Booking {booking_id: 'b1'}), (p:Payment {payment_id: 'p1'})
        CREATE (b)-[:PAID_BY {paid_at: '2024-12-01'}]->(p)
    """)
    session.run("""
        MATCH (cp:ChoreographerProfile {choreographer_id: 'cp1'}), (e:WorkshopEvent {event_id: 'e1'})
        CREATE (cp)-[:CONDUCTS]->(e)
    """)
    session.run("""
        MATCH (e:WorkshopEvent {event_id: 'e1'}),(s:Studio {studio_id: 's1'})
        CREATE (e)-[:HOSTED_AT {date: '2024-11-15', start_time: '10:00'}]->(s)
    """)
    print("Relationships Created")
````

#Graph Query

**What question do you want to answer and why does it matter to your
business?**

The business question I want to ask is *Which choreogrpahers are generting the most booking revenue and through which workshops?*

This matters because it tells who the top performers are, which styles are selling and where to invest.

````python
with driver.session() as session :
  result = session.run("""
      MATCH (cp:ChoreographerProfile)-[:CONDUCTS]->(e:WorkshopEvent)
              <-[:FOR_EVENT]-(b:Booking)
              -[:PAID_BY]->(p:Payment)

      WHERE p.payment_status = 'completed'
      RETURN cp.name AS choreographer,
             e.title AS workshop,
             count(b) AS total_bookings,
             sum(p.amount) AS total_revenue
     ORDER BY total_revenue DESC
    """)

  for row in result:
    print(f"{row['choreographer']} | {row['workshop']} |"
              f"bookings: {row['total_bookings']} | "
              f"revenue: £{row['total_revenue']:.2f}")
````

This query is a graph query as it traverses through 4 nodes and follows a natural connection between them. In relational database this query would require three JOINS across four tables.

The path reads a sentence "a choreographer conducts an event, which has bookings, which have payments."

````python
#clear all the existing pipelines to create new

gds.run_cypher("CALL gds.beta.pipeline.drop('lp_pipeline', false)")
````

````python
#create a new link prediction pipeline

gds.run_cypher("CALL gds.beta.pipeline.linkPrediction.create('lp_pipeline')")
````

````python
#Turn a user's network position into a 128-number vector using FASTRP

gds.run_cypher("""CALL gds.beta.pipeline.linkPrediction.addNodeProperty('lp_pipeline', 'fastRP',
                  { mutateProperty: 'embedding',
                    embeddingDimension: 128,
                    randomSeed: 42
                    })
               """)
````

````python
#combine vectors (multiplies User A's vector with User B's vector)

gds.run_cypher("""
CALL gds.beta.pipeline.linkPrediction.addFeature('lp_pipeline', 'hadamard',
                  { featureName: 'dotProduct',
                    nodeProperties: ['embedding']
                    })
               """)

print("pipeline built successfully")
````

##Preprocessing Data

````python
#1. Train/test split
gds.run_cypher(""" CALL gds.beta.pipeline.linkPrediction.configureSplit('lp_pipeline', {
                    testFraction:0.2,
                    trainFraction:0.8,
                    validationFOlds: 5
                    })
                    """)
````

````python
#add logistic regression

gds.run_cypher(""" CALL gds.beta.pipeline.linkPrediction.addLogisticRegression('lp_pipeline',
                {
                 maxEpochs:100,
                  penalty:0.01
                })
                """)
````

````python
#clear any existing models

gds.run_cypher("CALL gds.beta.model.drop('lastfm_lp_model', false)")
````

##Training Model

````python
# 4. Train the model and get the evaluation metrics

train_results = gds.run_cypher("""CALL gds.beta.pipeline.linkPrediction.train('lastfm',
                                {
                                  pipeline: 'lp_pipeline',
                                  modelName: 'lastfm_lp_model',
                                  targetRelationshipType: 'IS_FRIEND',
                                  randomSeed: 42
                                })
                                YIELD modelInfo
                                RETURN
                                  modelInfo.bestParameters AS best_parameters,
                                  modelInfo.modelName AS model,
                                  modelInfo.metrics.AUCPR.outerTrain AS train_aucpr,
                                  modelInfo.metrics.AUCPR.test AS test_aucpr,
                                  modelInfo.train.roc
                                """)

print("-----Training Results-----")
display(train_results)
````

##Predictions

We will look at users who are not currently connected and predict most likely top 10 new predictions

````python
# Make predictions
predictions = gds.run_cypher("""
CALL gds.beta.pipeline.linkPrediction.predict.stream('lastfm', {
  modelName: 'lastfm_lp_model',
  topN: 10,
  threshold: 0.5
})
YIELD node1, node2, probability
RETURN
  gds.util.nodeProperty('lastfm', node1, 'rawId') AS user1_id,
  gds.util.nodeProperty('lastfm', node2, 'rawId') AS user2_id,
  probability

ORDER BY probability DESC
""")

print("--- Top 10 Predicted New Friendships ---")
display(predictions)
````

## How did you do?

1. FASTRP algorythm : to generate the node embeddings, which translated each users positions in the social network as a vector - Users with similar friend circles received similar vectors.

2. Hadamard Product : Combine pairs of user vectors into single vector

3. Test train split into 80/20 and logistic regression classifier was used to predict the probability of the relation "IS_FRIEND" link existing based on those combined features

4. The model successfully trained and was evaluated using the AUCPR (Area Under the Precision-Recall Curve) metric.

Because social networks heavily rely on triadic closure (friends of friends become friends), the structural FastRP embeddings captured these graph patterns well, allowing the model to accurately predict the top 10 new likely connections.


#EXTRA CREDIT - Graph Algorythm

Using the PAGERANK algorythm to find most influencial users based on friendships



````python
pagerank_results = gds.run_cypher(""" CALL gds.pageRank.stream('lastfm', {
                    relationshipTypes: ['IS_FRIEND']
                    })
                    YIELD nodeId, score
                    RETURN
                     gds.util.nodeProperty('lastfm', nodeId, 'rawId') AS user_id,
                     score AS pagerank_score
                    ORDER BY pagerank_score DESC
                    LIMIT 10

                    """)
print("------TOP 10 MOST INFLUENCIAL USERS-------")
display(pagerank_results)
````

● Explain in words what you were trying to do / what your algorithm did.

I wanted to identify the most 'influential' or central users within the LastFM social network. Instead of simply counting which user has the highest raw number of friends (which is basic Degree Centrality). I wanted to find users who are deeply embedded in the network and connected to other highly connected users

I used the PageRank algorithm, which measures the transitive influence of nodes in a graph. It operates on the principle that connections are not created equal and a friendship with a highly influential user contributes more to your own influence score than a friendship with an isolated user.


Mathematicallly, it is a random surfer traversing through the network

The final pagerank score represents the probability that this random surfer will eventually land on that specific user.

The code streamed the graph data, ran this calculation across the IS_FRIEND relationships, and returned the top 10 users sorted by their PageRank score, revealing the core influencers of the LastFM community.

## Why I Chose the Topic and How It Relates to Graph Databases.

GraphRAG is the most exciting evolutions right now as traditional RAG systems rely on vector databases, which store textt chunks as flat isolated data points just based on semantice similarity.

GraphRAG shifts this foundation by using graph dataabses like neo4j to map out the data. GraphRAG uses the interconnected structure of graph databases to feed LLMs a map of relationships rather than just pile of text.


## Pretend You Are Teaching Someone About It

It works by first creating a map by extracting the entities and the relationships . And when the user asks a question, instead of searching for the documents, it literally traces the connected lines.

Take a real world use case for Financial Fraud Detection.Fraudsters often hide illicit money movements through layers of shell companies and micro-transactions.

Standard Graph, might pull documents about one suspicious company, but it misses the bigger picture.

On the other side, GraphRAG can traverse through the network - "COMPANY A TRANSFEERED TO COMPANY B, WHICH SHARES A DIRECTOR WITH COMPANY C AND WHICH WAS PREVIOUSLY FLAGGED FOR MONEY LAUNDERING " . This allow the AI to accurately explain the multi-hop money laundering scheme.


## Are There Any Tradeoffs or Pitfalls to Watch Out For?

Yes, Graph RAG is powerful, but it comes with several distinct challenges:
1. High Indexing Costs
2. Difficulty with dynamic Updates
3. Extraction Hallucinations
4. Overkill for Simple Queries

##Did You Learn Anything Particularly Interesting or Surprising?

One surprising fact is that Graph RAG despite adding layers , can actually lower the token costs and reduce hallucinations at query time. This is because the standardRAG lacks relationship awareness, it often has to stuff the LLM's context window with more context. GraphRAG on the other hand only lets in the exact entities and explicit relationships relevant to the prompt.
